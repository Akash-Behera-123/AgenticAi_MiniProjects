```python
!pip install -q pypdf

from pypdf import PdfReader
import os

documents = []

for filename in os.listdir(folder_path):
    if filename.lower().endswith(".pdf"):
        path = os.path.join(folder_path, filename)
        reader = PdfReader(path)

        for page in reader.pages:
            text = page.extract_text()

            if text:
                documents.append({
                    "text": text,
                    "source": filename
                })

print("PDFs:", len([f for f in os.listdir(folder_path) if f.endswith(".pdf")]))
print("Pages extracted:", len(documents))
```

```python
!pip install -q langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = []

for doc in documents:
    parts = splitter.split_text(doc["text"])

    for part in parts:
        chunks.append({
            "text": part,
            "source": doc["source"]
        })

print("Total chunks:", len(chunks))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded!")
```

```python
from langchain_community.vectorstores import FAISS

print("FAISS loaded!")
```

```python
test_chunks = chunks[:100]

texts = [chunk["text"] for chunk in test_chunks]
metadatas = [{"source": chunk["source"]} for chunk in test_chunks]

vectorstore = FAISS.from_texts(
    texts=texts,
    embedding=embeddings,
    metadatas=metadatas
)

print("Vector database created!")
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

query = "What is NIT Rourkela?"

results = retriever.invoke(query)

for i, doc in enumerate(results, 1):
    print(f"\n--- Result {i} ---")
    print("Source:", doc.metadata["source"])
    print(doc.page_content[:500])
```

```python
def ask_nitr(question):
    results = retriever.invoke(question)

    context = "\n\n".join(doc.page_content for doc in results)

    prompt = f"""
You are an intelligent NIT Rourkela assistant.

Use the documents when they contain relevant information.
If the documents are not relevant to the question, use your general knowledge.
If the question asks for current information, do not guess.

Conversation history:
{chat_history[-3:]}

Documents:
{context}

Question:
{question}

Answer:
"""

    response = model.invoke(prompt)
    answer = response.content

    chat_history.append((question, answer))

    # Only show sources when the retrieved context appears relevant
    sources = list(set(doc.metadata["source"] for doc in results))

    return answer, sources
```

```python
def web_search(question, max_results=3):
    results = DDGS().text(
        question,
        max_results=max_results
    )

    context = "\n\n".join(
        f"Title: {r['title']}\nURL: {r['href']}\nContent: {r['body']}"
        for r in results
    )

    return context

print(web_search("Director NIT Rourkela 2026"))
```
