```python
import requests
import google.auth
import langchain
import langchain_google_genai

print("requests:", requests.__version__)
print("google-auth:", google.auth.__version__)
print("langchain:", langchain.__version__)
print("Setup successful!")
```

```python
import os
from getpass import getpass

os.environ["GOOGLE_API_KEY"] = getpass("Enter your Gemini API key: ")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0
)

print("Gemini LLM created successfully!")
```

```python
response = llm.invoke(
    "Explain in one sentence what an AI agent is."
)

print(response.content)
```

```python
def get_weather(city):

    # 1. Find the city's coordinates
    geo_url = "https://geocoding-api.open-meteo.com/v1/search"

    geo_params = {
        "name": city,
        "count": 1,
        "language": "en",
        "format": "json"
    }

    geo_response = requests.get(
        geo_url,
        params=geo_params
    )

    geo_data = geo_response.json()

    # 2. Handle unknown city
    if "results" not in geo_data:
        return {
            "error": f"City '{city}' could not be found."
        }

    location = geo_data["results"][0]

    latitude = location["latitude"]
    longitude = location["longitude"]

    # 3. Get current weather
    weather_url = "https://api.open-meteo.com/v1/forecast"

    weather_params = {
        "latitude": latitude,
        "longitude": longitude,
        "current": [
            "temperature_2m",
            "relative_humidity_2m",
            "wind_speed_10m"
        ],
        "timezone": "auto"
    }

    weather_response = requests.get(
        weather_url,
        params=weather_params
    )

    weather_data = weather_response.json()

    current = weather_data["current"]

    # 4. Return only useful information
    return {
        "city": location["name"],
        "country": location.get("country"),
        "temperature": current["temperature_2m"],
        "humidity": current["relative_humidity_2m"],
        "wind_speed": current["wind_speed_10m"],
        "timezone": weather_data["timezone"]
    }
```

```python
result = get_weather("Mumbai")

print(result)
```

```python
from langchain_core.tools import tool

@tool
def weather_tool(city: str) -> dict:
    """
    Get the current weather for a city.
    Returns temperature, humidity, wind speed, country, and timezone.
    """
    return get_weather(city)

print("Tool name:", weather_tool.name)
print("Tool description:", weather_tool.description)
print("Tool schema:", weather_tool.args_schema.model_json_schema())
```

```python
llm_with_tools = llm.bind_tools([weather_tool])

print("Weather tool successfully connected to Gemini!")
```

```python
response = llm_with_tools.invoke(
    "What is the current weather in Mumbai? "
    "Use the weather tool to get the answer."
)

print("Gemini's response:")
print(response)

print("\nTool calls requested:")
print(response.tool_calls)
```

```python
tool_call = response.tool_calls[0]

print("Tool requested:", tool_call["name"])
print("Arguments:", tool_call["args"])

tool_result = weather_tool.invoke(tool_call["args"])

print("\nTool result:")
print(tool_result)
```

```python
from langchain_core.messages import ToolMessage

tool_message = ToolMessage(
    content=str(tool_result),
    tool_call_id=tool_call["id"]
)

final_response = llm_with_tools.invoke([
    {
        "role": "user",
        "content": "What is the current weather in Mumbai?"
    },
    response,
    tool_message
])

print(final_response.content)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[weather_tool],
    system_prompt=(
    "You are a weather assistant. "
    "Use weather_tool for current weather data. "
    "Keep answers short and structured. "
    "For comparisons, use:\n"
    "City 1: value\n"
    "City 2: value\n"
    "Result: city with higher value\n"
    "Do not add unnecessary explanations."
)
)

print("Autonomous weather agent created!")
```

```python
result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is the current weather in Mumbai?"
        }
    ]
})

print(result["messages"][-1].content)
```

```python
for i, message in enumerate(result["messages"]):
    print(f"\n--- Message {i} ---")
    print("Type:", type(message).__name__)
    print("Content:", message.content)

    if hasattr(message, "tool_calls") and message.tool_calls:
        print("Tool calls:", message.tool_calls)
```
